# this is correlationsship fuctions

```python
movies_pivot_table = pd.pivot_table(data=movies,values="rating",index="movie",columns="userId").fillna(0)
movies_pivot_table.columns =movies["userId"].unique()
movies_pivot_table
```

```python
movies_pivot_table.corr()
```

# we have another fuctions to use to find distance of the metrix

```python
euclidean_pivot_table = pd.pivot_table(data=movies,values="rating",index="userId",columns="movie").fillna(0)
euclidean_pivot_table.columns = movies["movie"].unique()
euclidean_pivot_table.index = movies["userId"].unique()
euclidean_pivot_table
```

```python
euclidean_corr = pd.DataFrame(pairwise_distances(X =euclidean_pivot_table,metric='euclidean',))
euclidean_corr.columns = movies["userId"].unique()
euclidean_corr.index = movies["userId"].unique()
euclidean_corr
```

# Building UBRF using cosine similarity

```python
euclidean_corr = pd.DataFrame(1 - pairwise_distances(X =euclidean_pivot_table,metric="cosine"))
euclidean_corr.columns = movies["userId"].unique()
euclidean_corr.index = movies["userId"].unique()
euclidean_corr
```
